## Importing, inspecting, cleaning and descriptive statistics

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

df = pd.read_csv("Nykka_dataset.csv")

print("Shape:", df.shape)
print("\nFirst five rows:")
display(df.head())
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
print("\nDataset information:")
df.info()
print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("Numerical summary:")
display(df.describe())
print("\nAll-column summary:")
display(df.describe(include="all").T)
```

## Data cleaning

```python
data = df.copy()

# Remove duplicate rows and records without a usable target.
data = data.drop_duplicates()
data["price"] = pd.to_numeric(data["price"], errors="coerce")
data["mrp"] = pd.to_numeric(data["mrp"], errors="coerce")
data = data.dropna(subset=["price", "mrp"])
data = data[(data["price"] > 0) & (data["mrp"] > 0)]

# Keep only valid/usable values for the target and core numerical fields.
data["rating"] = pd.to_numeric(data["rating"], errors="coerce")
data["rating_count"] = pd.to_numeric(data["rating_count"], errors="coerce")
data["rating"] = data["rating"].fillna(data["rating"].median())
data["rating_count"] = data["rating_count"].fillna(data["rating_count"].median())

# Missing categorical values are represented explicitly.
data["brand_name"] = data["brand_name"].fillna("Unknown").astype(str)
data["product_title"] = data["product_title"].fillna("Unknown").astype(str)
data["in_stock"] = data["in_stock"].fillna("Unknown").astype(str)

print("Cleaned shape:", data.shape)
print("\nMissing values after basic cleaning:")
display(data[["price", "mrp", "brand_name", "product_title", "rating", "rating_count", "in_stock"]].isnull().sum())
display(data.head())
```

## Descriptive statistics: mean, median, mode, range, variance, standard deviation, quartiles and IQR

```python
price = data["price"]
q1 = price.quantile(0.25)
q3 = price.quantile(0.75)

price_statistics = pd.Series({
    "Mean": price.mean(),
    "Median": price.median(),
    "Mode": price.mode().iloc[0],
    "Minimum": price.min(),
    "Maximum": price.max(),
    "Range": price.max() - price.min(),
    "Variance": price.var(),
    "Standard Deviation": price.std(),
    "Q1 (25%)": q1,
    "Q3 (75%)": q3,
    "IQR": q3 - q1
})
display(price_statistics.to_frame("Value"))
```

## Sampling distribution, point estimation and Central Limit Theorem

```python
np.random.seed(42)
population = data["price"].dropna().to_numpy()
sample_size = min(30, len(population))
number_of_samples = 1000
sample_means = [
    np.random.choice(population, size=sample_size, replace=True).mean()
    for _ in range(number_of_samples)
]

print("Population mean:", round(population.mean(), 2))
print("Mean of sample means:", round(np.mean(sample_means), 2))
print("Population standard deviation:", round(population.std(ddof=1), 2))
print("Standard deviation of sample means:", round(np.std(sample_means, ddof=1), 2))

plt.figure(figsize=(9, 5))
plt.hist(sample_means, bins=30, edgecolor="black")
plt.title(f"Sampling Distribution of Mean Price (n={sample_size}, {number_of_samples} samples)")
plt.xlabel("Sample mean price (₹)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Compare sample sizes to illustrate the Central Limit Theorem.
np.random.seed(42)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, n in zip(axes, [5, 30, 100]):
    means = [np.random.choice(population, size=n, replace=True).mean()
             for _ in range(1000)]
    ax.hist(means, bins=25, edgecolor="black")
    ax.set_title(f"Sample size = {n}")
    ax.set_xlabel("Sample mean price (₹)")
    ax.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## Confidence interval and hypothesis tests

```python
# 95% confidence interval for the mean selling price using a t distribution.
sample = data["price"].dropna()
n = len(sample)
sample_mean = sample.mean()
sample_std = sample.std(ddof=1)
t_critical = stats.t.ppf(0.975, df=n - 1)
margin_error = t_critical * sample_std / np.sqrt(n)
print("Sample mean price: ₹", round(sample_mean, 2))
print("95% confidence interval: ₹", round(sample_mean - margin_error, 2),
      "to ₹", round(sample_mean + margin_error, 2))
```

```python
# One-sample t-test: test whether the mean selling price differs from ₹500.
# ₹500 is a chosen reference value for this example, not a known population mean.
reference_price = 500
t_stat, p_value = stats.ttest_1samp(data["price"], popmean=reference_price)
print("One-sample t-test")
print("Reference mean: ₹", reference_price)
print("Sample mean: ₹", round(data["price"].mean(), 2))
print("t-statistic:", t_stat)
print("p-value:", p_value)
alpha = 0.05
print("Decision:", "Reject the null hypothesis" if p_value < alpha else "Fail to reject the null hypothesis")
```

```python
# Paired t-test: compare MRP and selling price for the same products.
t_stat, p_value = stats.ttest_rel(data["mrp"], data["price"])
print("Paired t-test: MRP vs selling price")
print("Average MRP: ₹", round(data["mrp"].mean(), 2))
print("Average selling price: ₹", round(data["price"].mean(), 2))
print("Average difference (MRP - price): ₹", round((data["mrp"] - data["price"]).mean(), 2))
print("t-statistic:", t_stat)
print("p-value:", p_value)
print("Decision:", "Reject the null hypothesis" if p_value < 0.05 else "Fail to reject the null hypothesis")
```

```python
# Independent two-sample t-test: compare prices between the two brands with the most products.
top_two_brands = data["brand_name"].value_counts().head(2).index.tolist()
group_a = data.loc[data["brand_name"] == top_two_brands[0], "price"].dropna()
group_b = data.loc[data["brand_name"] == top_two_brands[1], "price"].dropna()

t_stat, p_value = stats.ttest_ind(group_a, group_b, equal_var=False)
print("Brand A:", top_two_brands[0], "| mean price: ₹", round(group_a.mean(), 2))
print("Brand B:", top_two_brands[1], "| mean price: ₹", round(group_b.mean(), 2))
print("t-statistic:", t_stat)
print("p-value:", p_value)
print("Decision:", "Reject the null hypothesis" if p_value < 0.05 else "Fail to reject the null hypothesis")
```

```python
# One-way ANOVA: compare selling prices across the five brands with the most products.
top_five = data["brand_name"].value_counts().head(5).index
groups = [data.loc[data["brand_name"] == brand, "price"].dropna() for brand in top_five]
print("Brands:", list(top_five))
for brand, group in zip(top_five, groups):
    print(brand, "mean price: ₹", round(group.mean(), 2), "| count:", len(group))
f_stat, p_value = stats.f_oneway(*groups)
print("F-statistic:", f_stat)
print("p-value:", p_value)
print("Decision:", "Reject the null hypothesis" if p_value < 0.05 else "Fail to reject the null hypothesis")
```

```python
# Chi-square test of independence: brand and stock status.
# Use the top 10 brands to keep the contingency table readable.
top_ten = data["brand_name"].value_counts().head(10).index
chi_data = data[data["brand_name"].isin(top_ten)].copy()
contingency_table = pd.crosstab(chi_data["brand_name"], chi_data["in_stock"].astype(str))
display(contingency_table)
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)
print("Decision:", "Reject the null hypothesis" if p_value < 0.05 else "Fail to reject the null hypothesis")
```

## Regression models, model comparison and evaluation

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import joblib

# Use features available before predicting price. product_id and URLs are identifiers,
# and tags is omitted because most values are missing.
features = ["brand_name", "product_title", "mrp", "rating", "rating_count", "in_stock"]
target = "price"
model_data = data[features + [target]].copy()
X = model_data[features]
y = model_data[target]

numerical_features = ["mrp", "rating", "rating_count"]
categorical_features = ["brand_name", "in_stock"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

```python
# Preprocess numerical, categorical and text features.
# The preprocessing is fitted only on training data to avoid data leakage.
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]), numerical_features),
        ("categorical", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features),
        ("title_text", TfidfVectorizer(max_features=300, stop_words="english"), "product_title")
    ],
    remainder="drop"
)
print("Preprocessing created.")
```

```python
models = {
    "Multiple Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=0.01, max_iter=10000),
    "Decision Tree": DecisionTreeRegressor(max_depth=15, random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
}

trained_models = {}
predictions = {}
results = []

for name, estimator in models.items():
    model = Pipeline([
        ("preprocessor", preprocessor),
        ("model", estimator)
    ])
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    trained_models[name] = model
    predictions[name] = pred
    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "MSE": mean_squared_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R2 Score": r2_score(y_test, pred)
    })

results_df = pd.DataFrame(results).sort_values("R2 Score", ascending=False)
display(results_df)
```

```python
best_model_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]
best_predictions = predictions[best_model_name]

print("Best model based on test R2:", best_model_name)
print("MAE:", round(results_df.iloc[0]["MAE"], 2))
print("RMSE:", round(results_df.iloc[0]["RMSE"], 2))
print("R2 Score:", round(results_df.iloc[0]["R2 Score"], 4))
```

```python
plt.figure(figsize=(7, 6))
plt.scatter(y_test, best_predictions, alpha=0.6)
low = min(y_test.min(), best_predictions.min())
high = max(y_test.max(), best_predictions.max())
plt.plot([low, high], [low, high], linestyle="--")
plt.xlabel("Actual selling price (₹)")
plt.ylabel("Predicted selling price (₹)")
plt.title("Actual vs Predicted Prices - " + best_model_name)
plt.show()
```

## Gradio prediction interface

```python
# Run this cell only if Gradio is installed: pip install gradio
import gradio as gr

loaded_model = joblib.load("beauty_price_model.pkl")
brand_choices = sorted(data["brand_name"].dropna().astype(str).unique().tolist())
stock_choices = sorted(data["in_stock"].dropna().astype(str).unique().tolist())

def predict_beauty_price(brand_name, product_title, mrp, rating, rating_count, in_stock):
    input_df = pd.DataFrame([{
        "brand_name": brand_name,
        "product_title": product_title,
        "mrp": mrp,
        "rating": rating,
        "rating_count": rating_count,
        "in_stock": in_stock
    }])
    prediction = loaded_model.predict(input_df)[0]
    return f"Estimated selling price: ₹{prediction:,.2f}"

demo = gr.Interface(
    fn=predict_beauty_price,
    inputs=[
        gr.Dropdown(choices=brand_choices, label="Brand"),
        gr.Textbox(label="Product title"),
        gr.Number(value=float(data["mrp"].median()), label="MRP (₹)"),
        gr.Number(value=float(data["rating"].median()), label="Rating"),
        gr.Number(value=float(data["rating_count"].median()), label="Rating count"),
        gr.Dropdown(choices=stock_choices, label="In stock")
    ],
    outputs=gr.Textbox(label="Predicted price"),
    title="Beauty Product Price Prediction",
    description="Enter product details to estimate its selling price."
)
demo.launch()
```
